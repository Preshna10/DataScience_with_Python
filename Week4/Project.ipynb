### Descriptive Statistics
Descriptive statistics were calculated to summarize the main characteristics of the numerical variables.
The `describe()` function provides:

- Count
- Mean
- Standard deviation
- Minimum
- First quartile (25%)
- Median (50%)
- Third quartile (75%)

Additional statistics such as `mode and variance` were calculated separately, and `correlation` was used to examine relationships between numerical variables.

# Mean

```python
# Calculating the average alcohol level:
mean_alcohol = clean_df["alcohol"].mean()

print(
    "Mean alcohol:",
    round(mean_alcohol, 2)
)
```

### Mean
The mean represents the average value of a variable.

The mean alcohol measurement was calculated by adding all alcohol values and dividing by the number of observations.

# Median

```python
median_alcohol = clean_df["alcohol"].median()

print(
    "Median alcohol:",
    round(median_alcohol, 2)
)
```

### Median
The median represents the middle value when observations are arranged in order.

It is less affected by unusually high or low values than the mean.

# Mode

```python
mode_alcohol = clean_df["alcohol"].mode()

print("Mode alcohol:")
print(mode_alcohol)
```

```python
# We can also show only the first mode:
print(
    "First mode:",
    clean_df["alcohol"].mode()[0]
)
```

### Mode
The mode is the most frequently occurring value.

For continuous numerical variables such as alcohol, the mode may be less informative than the mean or median.

# Variance

```python
variance_alcohol = clean_df["alcohol"].var()

print(
    "Variance:",
    round(variance_alcohol, 2)
)
```

### Variance
Variance measures how spread out observations are around their mean.

A larger variance indicates greater variability in the data.

# Standard Deviation

```python
std_alcohol = clean_df["alcohol"].std()

print("Standard deviation:", round(std_alcohol, 2))
```

### Standard Deviation
Standard deviation describes the typical amount by which observations differ from the mean.

A smaller standard deviation means observations are more closely grouped around the mean.

# Putting important Statistics in One Table

```python
selected_columns = [
    "alcohol",
    "malic_acid",
    "magnesium",
    "color_intensity",
    "proline"
]
```

```python
# creating the table
statistics_summary = pd.DataFrame({
    "Mean": clean_df[selected_columns].mean(),
    "Median": clean_df[selected_columns].median(),
    "Mode": clean_df[selected_columns].mode().iloc[0],
    "Variance": clean_df[selected_columns].var(),
    "Standard Deviation": clean_df[selected_columns].std(),
    "Minimum": clean_df[selected_columns].min(),
    "Maximum": clean_df[selected_columns].max()
})

statistics_summary.round(2)
```

# Comparing Mean and Median

```python
print("Mean:", round(clean_df["alcohol"].mean(), 2))

print("Median:", round(clean_df["alcohol"].median(), 2))
```

# Correlation Analysis

```python
# First calculating correlation matrix:
correlation_matrix = clean_df.drop(columns=["target"]).corr()
correlation_matrix.round(2)
```

```python
# Focus on alcohol:
alcohol_correlation = (correlation_matrix["alcohol"].sort_values(ascending=False))
alcohol_correlation.round(2)
```

## Correlation Analysis

Correlation measures the strength and direction of the linear relationship between two numerical variables.

Correlation values range from -1 to +1.

- Values near +1 indicate a strong positive relationship.
- Values near -1 indicate a strong negative relationship.
- Values near 0 indicate a weak linear relationship.

# Alcohol and Proline Correlation

```python
alcohol_proline_corr = clean_df["alcohol"].corr(clean_df["proline"])

print("Correlation between alcohol and proline:",round(alcohol_proline_corr, 2))
```

# Comparing Wine Classes

```python
# Calculating averages by class:
class_summary = clean_df.groupby(
    "target"
)[
    [
        "alcohol",
        "malic_acid",
        "magnesium",
        "color_intensity",
        "proline"
    ]
].mean()

class_summary.round(2)
```

### Comparison by Wine Class

The dataset was grouped by the target variable to compare average chemical measurements among the three wine classes.

The results show that the wine classes differ in several measurements, including alcohol, color intensity, and proline.

# Class Counts

```python
class_counts = clean_df["target"].value_counts().sort_index()
class_counts
```
