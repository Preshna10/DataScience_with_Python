```python
v = {'a' , 'e' , 'i' , 'o' ,'u'}
cnt = 0
sent = sent.lower()
for c in sent:
    if c in v:
        cnt+=1
print(f"number of vowel character:{cnt}")
```

```python
sent = 'This is my book. I love my book.'
word = 0
for c in sent:
    if c.isspace():
        word+=1
print(f"number of words:{word+1}")
```

```python
#Count all the consonants in the sentence.
test = 'this is my computer, i love my computer. computers are very interesting device'
test
```

```python
v = {'a','e','i','o','u'}
cnt = 0
for c in test:
    if c not in v and c!=' ' and c!=',' and c!='.':
        cnt+=1
print(f"the number of consonants:{cnt}")
```
