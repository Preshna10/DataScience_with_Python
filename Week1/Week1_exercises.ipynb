# Lists

```python
#Q.No.5)Write a program to find: Highest mark, Lowest mark, Average mark, Number of students who scored ≥ 80, Number of students who failed (< 40).
#Student Marks: Given: marks = [78, 92, 65, 88, 55, 90, 73]

marks = [78, 92, 65, 88, 55, 90, 73]

# 1. Highest mark
highest = max(marks)

# 2. Lowest mark
lowest = min(marks)

# 3. Average mark
average = sum(marks) / len(marks)

# 4. Number of students who scored >= 80
scored_above_80 = len([m for m in marks if m >= 80])

# 5. Number of students who failed (< 40)
failed = len([m for m in marks if m < 40])

# Displaying results
print(f"Highest mark: {highest}")
print(f"Lowest mark: {lowest}")
print(f"Average mark: {average:.2f}")
print(f"Number of students who scored >= 80: {scored_above_80}")
print(f"Number of students who failed (< 40): {failed}")
```

```python
#Q.No.6) Remove Duplicates Given: numbers = [10, 20, 10, 30, 20, 40, 50, 30]
#Create a new list containing only unique values.

numbers = [10, 20, 10, 30, 20, 40, 50, 30]
unique_numbers = []
for num in numbers:
    if num not in unique_numbers:
        unique_numbers.append(num)
print(f"Unique numbers: {unique_numbers}")
```

```python
#Q.No.7) Find the Second Largest Number; numbers = [10, 50, 30, 50, 20, 40]
#Given a list of integers, find the second-largest unique number.

numbers = [10, 50, 30, 50, 20, 40]

largest = max(numbers)

remaining_numbers = [num for num in numbers if num != largest]

second_largest = max(remaining_numbers)

print(f"Largest number: {largest}")
print(f"Second largest unique number: {second_largest}")
```

```python
#Q13. Word Frequency Counter: Ask the user for a sentence and count how many times each word occurs.
sen = input("Enter the sentence: ")
sen = sen.lower()
words = sen.split()
frequency = {}
for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1
for word in frequency:
    print(word, "=", frequency[word])
```

```python
#Q.No.14) Employee Salary : Create a dictionary containing employee names and salaries: Find: Highest-paid employee, Lowest-paid employee, Average salary, Employees earning more than 40,000.
employees = {
    "Ram": 50000,
    "Sita": 35000,
    "Hari": 45000,
    "Gita": 30000,
    "John": 60000}

# Highest-paid employee
highest = max(employees, key=employees.get)
print("Highest-paid employee:", highest, employees[highest])

# Lowest-paid employee
lowest = min(employees, key=employees.get)
print("Lowest-paid employee:", lowest, employees[lowest])

# Average salary
average = sum(employees.values()) / len(employees)
print("Average salary:", average)

# Employees earning more than 40,000
print("Employees earning more than 40,000:")

for name, salary in employees.items():
    if salary > 40000:
        print(name, salary)
```

```python
#Q.No.16) Password Validator: Write a function: validate_password(password)
#A password is valid if it: contains at least 8 characters, contains at least one number, contains at least one uppercase letter, contains at least one lowercase letter.
def validate_password(password):
    if len(password) >= 8 and \
       any(c.isdigit() for c in password) and \
       any(c.isupper() for c in password) and \
       any(c.islower() for c in password):
        return "Valid Password"
    else:
        return "Invalid Password"


password = input("Enter your password: ")

print(validate_password(password))
```

```python
#Q.No.17) Student Grade Function: Create a function: calculate_grade(marks)
#Use:90–100 → A, 80–89  → B, 70–79  → C, 60–69  → D, Below 60 → F. Then ask the user for marks and display the grade.
def calculate_grade(marks):
    if marks < 0 or marks > 100:
        print("Invalid marks")
    elif marks >= 90:
        print("Grade: A")
    elif marks >= 80:
        print("Grade: B")
    elif marks >= 70:
        print("Grade: C")
    elif marks >= 60:
        print("Grade: D")
    else:
        print("Grade: F")

marks = float(input("Enter the marks: "))
calculate_grade(marks)
```

```python
#Q.No.18) Student File Reader: Create a text file called: students.txt
#Put several student names and marks inside it:Ram,85, Sita,92, Hari,76, Gita,88
#Write Python code to: Open the file, Read each line, Separate name and marks, Calculate the average mark, Print the student with the highest mark.
# Open the file
file = open("students.txt", "r")

total = 0
count = 0
highest_mark = 0
highest_student = ""

# Read each line
for line in file:
    line = line.strip()

    # Separate name and marks
    name, mark = line.split(",")

    mark = int(mark)

    print("Student:", name, "Marks:", mark)

    # Calculate total and count
    total += mark
    count += 1

    # Find highest mark
    if mark > highest_mark:
        highest_mark = mark
        highest_student = name

# Close the file
file.close()

# Calculate average
average = total / count

print("Average mark:", average)
print("Highest mark:", highest_student, highest_mark)
```

```python
#Q.No.19. File Word Counter: Create a program that accepts a filename and counts: number of lines, number of words, number of characters.
filename = input("Enter the filename: ")

file = open(filename, "r")

lines = 0
words = 0
characters = 0

for line in file:
    lines += 1
    words += len(line.split())
    characters += len(line)

file.close()

print("Number of lines:", lines)
print("Number of words:", words)
print("Number of characters:", characters)
```

# Basic OOP

```python
#Q.No.20) Bank Account Class: Create a class: class BankAccount:
#It should contain: Attributes, account holder name, account number, balance, Methods, deposit(amount), withdraw(amount), check_balance(), display_account().
class BankAccount:

    # Constructor
    def __init__(self, account_holder, account_number, balance):
        self.account_holder = account_holder
        self.account_number = account_number
        self.balance = balance

    # Deposit money
    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)

    # Withdraw money
    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient balance")

    # Check balance
    def check_balance(self):
        print("Current balance:", self.balance)

    # Display account details
    def display_account(self):
        print("Account Holder:", self.account_holder)
        print("Account Number:", self.account_number)
        print("Balance:", self.balance)


# Create an object
account = BankAccount("Ram", "12345", 10000)

account.display_account()

account.deposit(5000)

account.check_balance()

account.withdraw(3000)

account.check_balance()
```
