# Linear Regression
Linear Regression is a supervised machine learning algorithm used to predict numerical values. It finds the best-fitting linear relationship between the input features and target values by minimizing squared prediction errors.

## Import required libraries

```python
import numpy as np
```

```python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression
```

## Load the dataset

```python
diabetes = load_diabetes()

X_lr = diabetes.data
y_lr = diabetes.target
```

X_lr contains the input features, and y_lr contains the numerical values we want to predict.

```python
X_lr.shape
```

```python
y_lr.shape
```

```python
#Import the function to split the dataset into training and testing sets

from sklearn.model_selection import train_test_split
```

## Split the dataset

```python
X_train_lr, X_test_lr, y_train_lr, y_test_lr = train_test_split(X_lr, y_lr, test_size=0.2, random_state=42)
```

We use 80% of the data for training and 20% for testing.

## Train Linear Regression

```python
#create model
linear_model = LinearRegression()
```

```python
#train the model
linear_model.fit(X_train_lr, y_train_lr)
```

## Make predictions

```python
y_pred_lr = linear_model.predict(X_test_lr)
print("First 5 predicted values:", np.round(y_pred_lr[:5], 2))
```

## Evaluate the model

```python
from sklearn.metrics import (mean_absolute_error, mean_squared_error, r2_score)
```

```python
lr_mae = mean_absolute_error(y_test_lr, y_pred_lr)

lr_mse = mean_squared_error(y_test_lr, y_pred_lr)

lr_rmse = np.sqrt(lr_mse)

lr_r2 = r2_score(y_test_lr, y_pred_lr)

print(f"MAE: {lr_mae:.2f}")
print(f"MSE: {lr_mse:.2f}")
print(f"RMSE: {lr_rmse:.2f}")
print(f"R²: {lr_r2:.3f}")
```

### Understanding the metrics
MAE: Average absolute difference between actual and predicted values.

MSE: Average squared difference between actual and predicted values.

RMSE: Square root of MSE, showing prediction errors in the target's original units.

R²: Measures how much variation in the target is explained by the model.

For MAE, MSE, and RMSE, smaller values are better. For R², values closer to 1 are generally better.

# Logistic Regression
Logistic Regression is a supervised machine learning algorithm used for classification. It predicts the probability of an observation belonging to a class. In this project, it is used to classify tumors as malignant or benign.

## Import required libraries

```python
from sklearn.datasets import load_breast_cancer
from sklearn.linear_model import LogisticRegression
```

## Load the dataset

```python
cancer = load_breast_cancer()

X_log = cancer.data

# 1 = malignant, 0 = benign
y_log = (cancer.target == 0).astype(int)
```

```python
X_log.shape
```

```python
y_log.shape
```

```python
#Import the function to split the dataset into training and testing sets

from sklearn.model_selection import train_test_split
```

## Split the dataset

```python
X_train_log, X_test_log, y_train_log, y_test_log = train_test_split(X_log, y_log, test_size=0.2, random_state=42)
```

## Train Logistic Regression

```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
```

```python
#create model
logistic_model = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=1000, random_state=42)
)
```

StandardScaler scales the features using the training data to help Logistic Regression train effectively.

```python
#train the model
logistic_model.fit(X_train_log, y_train_log)
```

## Make predictions

```python
y_pred_log = logistic_model.predict(X_test_log)
print("First 10 predicted classes:", y_pred_log[:10])
```

## Evaluate the model

```python
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score)
```

```python
log_accuracy = accuracy_score(y_test_log, y_pred_log)

log_precision = precision_score(y_test_log, y_pred_log)

log_recall = recall_score(y_test_log, y_pred_log)

log_f1 = f1_score(y_test_log, y_pred_log)

print(f"Accuracy: {log_accuracy:.2%}")
print(f"Precision: {log_precision:.2%}")
print(f"Recall: {log_recall:.2%}")
print(f"F1-score: {log_f1:.2%}")
```

### Understanding the metrics
Accuracy: Percentage of all predictions that are correct.

Precision: Of the tumors predicted malignant, how many were actually malignant.

Recall: Of the actually malignant tumors, how many were correctly identified.

F1-score: Balances precision and recall.
