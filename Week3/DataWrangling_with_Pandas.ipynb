```python
import pandas as pd
```

# Load CSV file

```python
data = r"D:\DataScience_with_Python\Week3\titanic.csv"

df = pd.read_csv(data)

print("Dataset loaded successfully!")
print(df.head())
```

# Exploring the dataset structure

```python
print("Shape of dataset:")
print(df.shape)

print("\nColumn names:")
print(df.columns)

print("\nData types:")
print(df.dtypes)

print("\nDataset information:")
print(df.info())

print("\nStatistical summary:")
print(df.describe())
```

# Checking missing values

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

# Handling missing values

```python
# Fill missing Age values with the median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing Embarked values with the mode
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# Drop Cabin because it contains many missing values
df = df.drop(columns=["Cabin"])

print("Missing values after cleaning:")
print(df.isnull().sum())
```

# Checking and removing duplicates

```python
print("Number of duplicate rows:")
print(df.duplicated().sum())

df = df.drop_duplicates()

print("Shape after removing duplicates:")
print(df.shape)
```

# Filtering rows

```python
# Filtering passengers older than 30.
older_passengers = df[df["Age"] > 30]

print("Passengers older than 30:")
print(older_passengers.head())
```

```python
# Filtering female passengers
female_passengers = df[df["Sex"] == "female"]

print(female_passengers.head())
```

```python
# Filtering passengers who survived
survivors = df[df["Survived"] == 1]

print(survivors.head())
```

```python
#combining conditions including those who are female and those who survived.
female_survivors = df[
    (df["Sex"] == "female") &
    (df["Survived"] == 1)
]

print(female_survivors.head())
```

# Practicing groupby

```python
# Average age by sex
avg_age_by_sex = df.groupby("Sex")["Age"].mean()

print("Average age by sex:")
print(avg_age_by_sex)
```

```python
# Survival rate by passenger class
survival_by_class = df.groupby("Pclass")["Survived"].mean()

print("Survival rate by passenger class:")
print(survival_by_class)
```

```python
# Count passengers by embarkation location
passengers_by_port = df.groupby("Embarked").size()

print("Passengers by embarkation port:")
print(passengers_by_port)
```

# Practicing merging

```python
# Creating another small dataset containing passenger-class descriptions.
class_info = pd.DataFrame({
    "Pclass": [1, 2, 3],
    "ClassName": [
        "First Class",
        "Second Class",
        "Third Class"
    ]
})

print(class_info)
```

```python
# Merging it with the Titanic dataset
merged_df = pd.merge(
    df,
    class_info,
    on="Pclass",
    how="left"
)

print("Merged dataset:")
print(merged_df.head())
```

# Creating an analysis-ready dataset

```python
# Removing columns that are not useful for a simple analysis.
clean_df = merged_df.drop(
    columns=[
        "PassengerId",
        "Ticket"
    ]
)

print(clean_df.head())
```

```python
# Create FamilySize by adding siblings/spouse, parents/children, and the passenger.
# Creating a useful new feature; SibSp= siblings or spouse, Parch= parents or childrens , 1= passenger themself
clean_df["FamilySize"] = (
    clean_df["SibSp"] +
    clean_df["Parch"] +
    1
)

print(clean_df[["Name", "FamilySize"]].head())
```

# Making column names cleaner

```python
clean_df.columns = clean_df.columns.str.lower()

print(clean_df.columns)
```

# Final checking

```python
print("Final dataset shape:")
print(clean_df.shape)

print("\nMissing values:")
print(clean_df.isnull().sum())

print("\nDuplicates:")
print(clean_df.duplicated().sum())

print("\nFinal data types:")
print(clean_df.dtypes)
```

```python
# Display the final cleaned dataset
clean_df.head()
```

# Saving the cleaned dataset

```python
clean_df.to_csv(
    "titanic_analysis_ready.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
